```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("StudentsPerformance_modified.csv")
```

```python
df.columns
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
df['math score'] = pd.to_numeric(df['math score'], errors='coerce')
```

```python
df['math score']
```

```python
# Fill original missing values with column means
for col in ['math score', 'reading score', 'writing score']:
    df[col].fillna(df[col].mean(), inplace=True)
```

```python
df.dtypes
```

```python
# Define outlier removal (IQR method)
def remove_outliers(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df[col] = df[col].where((df[col] >= lower) & (df[col] <= upper), np.nan)
    
```

```python
for col in ['math score', 'reading score', 'writing score']:
    remove_outliers(df, col)
    df[col].fillna(df[col].mean(), inplace=True)
```

```python
df.isna().sum()
```

```python
df['math score log'] = np.log1p(df['math score'])  
```

```python
# Plot before vs after log transform
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(df['math score'], kde=True)
plt.title('Original Math Score')

plt.subplot(1, 2, 2)
sns.histplot(df['math score log'], kde=True)
plt.title('Log-Transformed Math Score')

plt.tight_layout()
plt.show()
```
