```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag,word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
doc = '''Rajgad (literal meaning Ruling Fort) is a hill fort situated in the Pune district of
Maharashtra, India. Formerly known as Murumdev, the fort was the capital of the Mara
tha Empire under the rule of Chatrapati Shivaji Maharaj for almost 26 years, after w
hich the capital was moved to the Raigad Fort. Treasures discovered from an adjac
ent fort called Torna were used to completely build and fortify the Rajgad Fort.'''
```

```python
doc
```

```python
tok = word_tokenize(doc)
tok
```

```python
import string
```

```python
print(string.__doc__)
```

```python
string.punctuation
```

```python
tokens_real = [w for w in tok if w not in string.punctuation]
tokens_real
```

```python
pos = pos_tag(tokens_real)
pos
```

```python
stop_word = set(stopwords.words('english'))
```

```python
filtered_tokens = [w for w in tokens_real if w.lower() not in stop_word]
```

```python
print("After Stopwords Removal:", filtered_tokens)
```

```python
stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in filtered_tokens]
stemmed
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in filtered_tokens]
lemmatized
```

```python
corpus = tokens_real
vectorizer = TfidfVectorizer()
tf = vectorizer.fit_transform(corpus)
print("\nTF-IDF Matrix:")
print(tf.toarray())
```

```python
print("Feature Names:", vectorizer.get_feature_names_out())
```
