```python
import nltk
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer ,  WordNetLemmatizer
from nltk import pos_tag,word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
doc = '''Rajgad (literal meaning Ruling Fort) is a hill fort situated in the Pune district of
Maharashtra, India. Formerly known as Murumdev, the fort was the capital of the Mara
tha Empire under the rule of Chatrapati Shivaji Maharaj for almost 26 years, after w
hich the capital was moved to the Raigad Fort. Treasures discovered from an adjac
ent fort called Torna were used to completely build and fortify the Rajgad Fort.'''
```

```python
doc
```

```python
import string
```

```python
token = word_tokenize(doc)
token
```

```python
token_real = [w for w in token if  w not in string.punctuation]
token_real
```

```python
pos =  pos_tag(token_real)
pos
```

```python
stop_word = set(stopwords.words('english'))
stop_word
```

```python
filtered_tokens = [w for w in token_real if w not in stop_word]
filtered_tokens
```

```python
stemmer = PorterStemmer()
stemmed =  [stemmer.stem(w) for w in filtered_tokens]
stemmed
```

```python
print(stemmer.stem('working'))
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in filtered_tokens]
lemmatized
```

```python
corpus = [' '.join(filtered_tokens)]
vectorizer = TfidfVectorizer()
tf = vectorizer.fit_transform(corpus)
print("Matrix \n")
print(tf.toarray())
```
