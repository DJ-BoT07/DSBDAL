```python
import seaborn as sb
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = sb.load_dataset('titanic')
df
```

```python
sb.displot(df["fare"])
```

```python
sb.displot(df["fare"],kde=True, bins=10 )
```

```python
sb.jointplot(x="age", y="fare",data=df,hex="survive)
```

```python
sb.jointplot(x='age', y='fare', data=df, kind="hex")
```

```python
df=df.dropna()
sb.pairplot(df, hue='sex')
```

```python
plt.figure(figsize=(10,4))
sb.histplot(df["fare"],kde=True, bins=30, color="Blue")
plt.xlabel("Fare")
plt.ylabel("Count")
```
